```python
dbutils.widgets.text("group_name_run", "")
grp_run = dbutils.widgets.get("group_name_run")
print(grp_run)


dbutils.widgets.text("group_name_manage", "")
grp_manage = dbutils.widgets.get("group_name_manage")
print(grp_manage)
```

```python
from databricks.sdk import WorkspaceClient
from databricks.sdk.service import iam

w = WorkspaceClient()

w.permissions.update(
    request_object_type="notebooks",  # or "clusters", "jobs", "dashboards", etc.
    request_object_id="3238537659586702",
    access_control_list=[
        iam.AccessControlRequest(
            group_name=f"{grp_run}",
            permission_level=iam.PermissionLevel.CAN_RUN,
        ),
        iam.AccessControlRequest(
            group_name=f"{grp_manage}",
            permission_level=iam.PermissionLevel.CAN_MANAGE,
        ),
    ],
)
```
